```python
# Task 2 VOTES ANALYSIS
# identify the restaurants with the highest & lowest number of votes

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

#load the dataset
df = pd.read_csv(r"C:\Users\Acer\Downloads\Dataset.csv")
df.shape
```

```python
df = df.dropna(subset = ['Votes'])
# find restaurants with highest & lowest votes
most_voted = df.loc[df['Votes'].idxmax()]
least_voted = df.loc[df['Votes'].idxmax()]
print('Restaurant with most votes:',most_voted[['Restaurant Name','Votes']])
print('Restaurant with least votes:',least_voted[['Restaurant Name','Votes']])
```

```python
# analyze if there is a correlation between the number of votes and the rating of a restaurant

df['Votes'] = pd.to_numeric(df['Votes'],errors = 'coerce')
df['Aggregate rating'] = pd.to_numeric(df['Aggregate rating'],errors = 'coerce')
df_clean = df.dropna(subset = ['Votes','Aggregate rating'])
# correlation coefficient
correlation = df_clean['Votes'].corr(df_clean['Aggregate rating'])
print(f" Correlation between votes and rating: {correlation:.3f}")
```

```python
plt.figure(figsize = (8,5))
sns.regplot(x = 'Aggregate rating',y = 'Votes',data =df_clean,scatter_kws = {'alpha':0.5},line_kws = {'alpha':0.5})
plt.title("Votes vs Aggregate rating")
plt.xlabel("Aggregate rating")
plt.ylabel('Votes')
plt.grid(True)
plt.show()
```
